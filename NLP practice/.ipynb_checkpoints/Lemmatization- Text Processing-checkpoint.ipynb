# Wordnet Lemmatizer

**WordNet Lemmatizer** is a Natural Language Processing (NLP) tool from Python's NLTK library that reduces words to their meaningful base or dictionary form, known as a **lemma**.

Unlike stemming, which simply chops off word endings and can create non-words, lemmatization uses the [WordNet](https://wordnet.princeton.edu/) lexical database to consider context and part of speech. This ensures the output is always a real, valid word—for example, turning **"running"** into **"run"** and **"better"** into **"good"**. It helps computers group different forms of the same word together for accurate text analysis.

### use cases Q&A, chatbots, text summmarization

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer_ = WordNetLemmatizer()
```

pos-
Noun - n
Verb - v
Adjective - a
adverb - r

lemmatizer_.lemmatize("wording", pos='n')

```python
'''
pos-
Noun - n
Verb - v
Adjective - a
adverb - r
'''

print("going ---> (noun) -----> " +lemmatizer_.lemmatize("going", pos='n'))
print("going ---> (verb) -----> " +lemmatizer_.lemmatize("going", pos='v'))
print("going ---> (adjective) -----> " +lemmatizer_.lemmatize("going", pos='a'))
print("going ---> (adverb) -----> " +lemmatizer_.lemmatize("going", pos='r'))
```

```python
words = [
    "running", "runs", "ran", "runner", "easily", "fairly",
    "studies", "studying", "studied", "happy", "happier", "happiness",
    "generously", "generation", "general", "generally",
    "organization", "organize", "organizing", "organized",
    "programming", "programmer", "programs", "programmed",
    "better", "best", "good", "wolves", "flies", "flying",
    "history", "historical"
]
```

```python
for i in words:
    print(i + " ----> " + lemmatizer_.lemmatize(i, pos='n'))
```

```python
for i in words:
    print(i + " ----> " + lemmatizer_.lemmatize(i, pos='v'))
```

```python
for i in words:
    print(i + " ----> " + lemmatizer_.lemmatize(i, pos='a'))
```

```python
for i in words:
    print(i + " ----> " + lemmatizer_.lemmatize(i, pos='r'))
```

```python
lemmatizer_.lemmatize("goes", pos='n'),lemmatizer_.lemmatize("goes", pos='v'),lemmatizer_.lemmatize("goes", pos='a'),lemmatizer_.lemmatize("goes", pos='r')
```
