# 📌 Unit 5: Stemming Techniques in NLP (Text Normalization)

> **What comes after Tokenization?**  
> In Unit 4, we successfully converted paragraphs into sentences (Documents) and sentences into words (Tokens).  
> However, words like **"running"**, **"runs"**, and **"ran"** all carry the same core conceptual meaning.  
> **Stemming** is the rule-based process of slicing off affixes (prefixes and suffixes) to reduce words down to their **root / stem form**.

---

### 🗺️ Visual NLP Preprocessing Workflow:
```text
┌────────────────────────────────────────────────────────────────────────┐
│ [1. Raw Corpus] "Running, ran, and runs all share the same root word!" │
└───────────────────────────────────┬────────────────────────────────────┘
                                    │ sent_tokenize()
                                    ▼
┌────────────────────────────────────────────────────────────────────────┐
│ [2. Documents]  "Running, ran, and runs all share the same root word!" │
└───────────────────────────────────┬────────────────────────────────────┘
                                    │ word_tokenize()
                                    ▼
┌────────────────────────────────────────────────────────────────────────┐
│ [3. Tokens]     ['Running', 'ran', 'and', 'runs', 'all', 'share', ...] │
└───────────────────────────────────┬────────────────────────────────────┘
                                    │ Stemming (Porter / Snowball)
                                    ▼
┌────────────────────────────────────────────────────────────────────────┐
│ [4. Stems]      ['run', 'ran', 'and', 'run', 'all', 'share', ...]      │
└────────────────────────────────────────────────────────────────────────┘
```

---

### 📦 What is inside Unit 5?
1. **`PorterStemmer`**: The classic, fastest heuristic suffix-stripping algorithm.
2. **`RegexpStemmer`**: Custom pattern-based regex stemmer for specific suffixes.
3. **`SnowballStemmer`**: Also known as **Porter2** — improved accuracy and multilingual support.
4. **Visual Comparison:** Stemming vs. Lemmatization (Heuristic speed vs. Dictionary precision).

````python
## 1. PorterStemmer
from nltk.stem import PorterStemmer

porter = PorterStemmer()

words = ["eating", "eats", "eaten", "writing", "writes", "written", "programming", "programs", "finally", "finalized"]

for word in words:
    print(f"{word:15} ---> {porter.stem(word)}")
````

### ⚠️ Important Note on Stemming Limitations
Notice that Stemming applies **fixed grammatical rules** without checking a dictionary.  
Sometimes the resulting stem is **not a valid English word**:
- `congratulations` ➔ `congratul`
- `history` / `historical` ➔ `histori`
- `studies` ➔ `studi`

*(Later in Lemmatization, we will see how dictionaries solve this!)*

````python
## Stemming tokens inside our corpus documents
for sentence in documents:
    words_in_sent = word_tokenize(sentence)
    stemmed_words = [porter.stem(w) for w in words_in_sent]
    print("Original:", sentence)
    print("Stemmed :", " ".join(stemmed_words))
    print("-" * 75)
````

---
### 🛠️ 2. RegexpStemmer (Custom Regular Expression Stemmer)
Allows custom regex patterns to strip specific suffixes like `ing`, `ly`, `able`, or `ed` with a minimum word length threshold.

````python
from nltk.stem import RegexpStemmer

## Define custom regex rule to strip suffixes 'ing', 's', 'e', 'able', 'ly'
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$|ly$', min=4)

sample_words = ["eating", "flying", "adjustable", "cars", "happily", "reading"]

for w in sample_words:
    print(f"{w:15} ---> {reg_stemmer.stem(w)}")
````

---
### ❄️ 3. SnowballStemmer (Porter2 Algorithm)
- Developed by Martin Porter as a major improvement over PorterStemmer.
- Better handles edge cases, irregular suffixes, and vowel-consonant pairs.
- Supports **multiple languages** (English, French, Spanish, German, etc.).

````python
from nltk.stem import SnowballStemmer

snowball = SnowballStemmer('english')

comparison_words = ["generously", "fairly", "troubled", "sporting", "finally", "historical"]

print(f"{'Original Word':16} | {'PorterStemmer':16} | {'SnowballStemmer':16}")
print("-" * 55)
for w in comparison_words:
    print(f"{w:16} | {porter.stem(w):16} | {snowball.stem(w):16}")
````

---
### 📊 Visual Comparison: PorterStemmer vs. SnowballStemmer vs. WordNetLemmatizer

| Target Word | PorterStemmer (Rule-based) | SnowballStemmer (Porter2) | WordNetLemmatizer (Dictionary) | Valid English Root? |
| :--- | :--- | :--- | :--- | :---: |
| **eating** | `eat` | `eat` | `eat` | ✅ Yes |
| **studies** | `studi` | `studi` | `study` | ✅ Yes (Lemmatizer wins) |
| **fairly** | `fairli` | `fair` | `fairly` | ✅ Yes |
| **better** | `better` | `better` | `good` (pos='a') | ✅ Yes (Lemmatizer wins) |
| **corpora** | `corpora` | `corpora` | `corpus` (pos='n') | ✅ Yes (Lemmatizer wins) |
| **historical**| `histor` | `histori` | `historical` | ✅ Yes |

> 💡 **Key Takeaways for Your NLP Toolkit:**
> - **Stemming** is **faster** and best for information retrieval, search query indexing, and spam classification.
> - **Lemmatization** is **linguistically accurate** and best for question answering, chatbots, and sentiment analysis.

````python
## Preview of Next Step: Comparing Stemming vs Lemmatization
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words_to_compare = ["studies", "studying", "better", "corpora", "wolves", "crying"]

print(f"{'Word':12} | {'PorterStemmer':15} | {'SnowballStemmer':15} | {'Lemmatizer':15}")
print("=" * 65)
for w in words_to_compare:
    p_stem = porter.stem(w)
    s_stem = snowball.stem(w)
    lemma = lemmatizer.lemmatize(w, pos='v' if w in ['studies', 'studying', 'crying'] else 'n')
    if w == 'better':
        lemma = lemmatizer.lemmatize(w, pos='a')
    print(f"{w:12} | {p_stem:15} | {s_stem:15} | {lemma:15}")
````
