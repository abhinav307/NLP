# Wordnet Lemmatizer

**WordNet Lemmatizer** is a Natural Language Processing (NLP) tool from Python's NLTK library that reduces words to their meaningful base or dictionary form, known as a **lemma**.

Unlike stemming, which simply chops off word endings and can create non-words, lemmatization uses the [WordNet](https://wordnet.princeton.edu/) lexical database to consider context and part of speech. This ensures the output is always a real, valid word—for example, turning **"running"** into **"run"** and **"better"** into **"good"**. It helps computers group different forms of the same word together for accurate text analysis.
