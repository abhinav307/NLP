# Stemming

**Stemming** is a text preprocessing technique in Natural Language Processing (NLP) that reduces inflected or derived words to their base or root form—called a **stem**—by chopping off prefixes or suffixes using heuristic rules, even if the resulting stem is not a valid dictionary word (e.g., turning *"running,"* *"runs,"* and *"ran"* into *"run"*, or *"studies"* into *"studi"*).

```python
words = [
    "running", "runs", "ran", "runner", "easily", "fairly",
    "studies", "studying", "studied", "happy", "happier", "happiness",
    "generously", "generation", "general", "generally",
    "organization", "organize", "organizing", "organized",
    "programming", "programmer", "programs", "programmed",
    "better", "best", "good", "wolves", "flies", "flying",
    "history", "historical"
]
```

### PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for i in words:
    print(i+"----->"+ stemming.stem(i))
```

```python
stemming.stem('congratulations') #😒
```

```python
stemming.stem('sitting') #😘
```

# RegexpStemmer class

**RegexpStemmer** is a customizable stemming class in NLTK that takes a user-defined regular expression pattern and strips any matching prefix or suffix from a word, giving you direct rule-based control over how words are trimmed (for example, using `RegexpStemmer('ing$|s$|e$|able$', min=4)` to strip specific endings only from words with at least four characters).

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer= RegexpStemmer('ing' ,  min=4)
word1 = reg_stemmer.stem('ingingeatinging')
print(word1)
```

```python
reg_stemmer= RegexpStemmer('ing' ,  min=4)
word1 = reg_stemmer.stem('ingingeatingingmen')
print(word1)
```

```python
reg_stemmer= RegexpStemmer('ing' ,  min=4)
word1 = reg_stemmer.stem('ingeatingmening maningg')
print(word1)
```

```python
reg_stemmer= RegexpStemmer('ing$' ,  min=4)
word1 = reg_stemmer.stem('ingeatingmening')
print(word1)
```

```python
rex_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)

# Test words for each suffix rule + min=4 threshold + edge cases
print("eating   ---->", rex_stemmer.stem('eating'))      # strips 'ing'  -> 'eat'
print("cats     ---->", rex_stemmer.stem('cats'))        # strips 's'    -> 'cat'
print("love     ---->", rex_stemmer.stem('love'))        # strips 'e'    -> 'lov' --  😒😒😒
print("readable ---->", rex_stemmer.stem('readable'),"\n")    # strips 'able' -> 'read'

# Testing min=4 (words shorter than 4 chars are skipped)
print("see      ---->", rex_stemmer.stem('see'))         # len 3 (< 4)   -> 'see' (unchanged)
print("runs     ---->", rex_stemmer.stem('runs'),"\n")        # len 4 (>= 4)  -> 'run'

# Edge cases to watch out for
print("running  ---->", rex_stemmer.stem('running'))     # strips 'ing'  -> 'runn' (double consonant remains) --  😒😒😒
print("table    ---->", rex_stemmer.stem('table'))       # strips 'able' -> 't' (over-stemmed!) --  😒😒😒
print("sing     ---->", rex_stemmer.stem('sing'))        # len 4, ends in 'ing' -> 's' --  😒😒😒
```

# Snowball Stemmer

**Snowball Stemmer** (also known as **Porter2**) is an improved, faster, and more linguistically accurate version of the classic Porter Stemmer in NLTK that supports multiple languages (such as English, French, German, and Spanish) and handles certain suffixes—like `"fairly"` and `"sportingly"`—more cleanly than Porter without over-stemming as aggressively.
