## 🥊 Unit 1 (Practice Lab): The Tokenizer Benchmark Arena
Let's see how `word_tokenize`, `wordpunct_tokenize`, and `TreebankWordTokenizer` behave when faced with real-world complexities:
- **Contractions**: `can't`, `isn't`, `it's`
- **Punctuation & Currency**: `$49.99/mo!`, `100%`
- **Abbreviations**: `Dr. Smith`

````python
# Unit 1 Practice: Benchmark Tokenizers on Tricky Edge Cases
from nltk.tokenize import sent_tokenize, word_tokenize, wordpunct_tokenize, TreebankWordTokenizer

sample_text = "Dr. Smith said: 'NLP isn't easy in 2026, but it's 100% worth it at $49.99/mo!' Can't wait."

tb_tokenizer = TreebankWordTokenizer()

tok_word = word_tokenize(sample_text)
tok_punct = wordpunct_tokenize(sample_text)
tok_tb = tb_tokenizer.tokenize(sample_text)

print("=" * 85)
print("🥊 TOKENIZER BENCHMARK ON TRICKY TEXT:")
print("Sample:", sample_text)
print("=" * 85)
print(f"1. word_tokenize()       ({len(tok_word)} tokens): {tok_word[:8]}...")
print(f"2. wordpunct_tokenize()  ({len(tok_punct)} tokens): {tok_punct[:8]}...")
print(f"3. TreebankWordTokenizer ({len(tok_tb)} tokens): {tok_tb[:8]}...")
print("=" * 85)
print(f"{'Token Index':<12} | {'word_tokenize':<22} | {'wordpunct_tokenize':<22} | {'TreebankTokenizer':<22}")
print("-" * 85)
for i in range(min(12, max(len(tok_word), len(tok_punct), len(tok_tb)))):
    w1 = tok_word[i] if i < len(tok_word) else "-"
    w2 = tok_punct[i] if i < len(tok_punct) else "-"
    w3 = tok_tb[i] if i < len(tok_tb) else "-"
    print(f"Token #{i:<5} | {w1:<22} | {w2:<22} | {w3:<22}")
print("=" * 85)
````

## ✂️ Unit 2: Stemming and its Types (Text Processing & Suffix Stripping)
Stemming is a rule-based heuristic technique that cuts off prefixes or suffixes to reduce words to their base stem.
We compare the 3 primary stemmers in NLTK:
1. **PorterStemmer**: Original 1980 algorithm, fast, heuristic suffix chopping.
2. **RegexpStemmer**: Regex-driven custom stemmer tailored for specific business domains.
3. **SnowballStemmer (Porter2)**: Improved multi-language stemmer with better morphological handling.

````python
# Unit 2 Practice: Stemming Algorithms Showdown
from nltk.stem import PorterStemmer, RegexpStemmer, SnowballStemmer

porter = PorterStemmer()
snowball = SnowballStemmer('english')
regexp = RegexpStemmer('ing$|s$|ed$|ly$|ation$', min=4)

words_to_stem = [
    "running", "runs", "ran", "runner", "easily", "fairly",
    "studies", "studying", "cats", "generously", "multiply",
    "provision", "organization", "congratulations"
]

print("=" * 85)
print(f"{'Word':<16} | {'PorterStemmer':<18} | {'SnowballStemmer':<20} | {'RegexpStemmer':<18}")
print("=" * 85)
for w in words_to_stem:
    p = porter.stem(w)
    s = snowball.stem(w)
    r = regexp.stem(w)
    print(f"{w:<16} | {p:<18} | {s:<20} | {r:<18}")
print("=" * 85)
````

## 🧹 Unit 3: Stopwords Removal & Noise Reduction Lab
Stopwords (`the`, `is`, `at`, `which`) are extremely frequent words that add little contextual significance in search and topic classification.
Removing them reduces dataset size significantly and brings out the core topics!

````python
# Unit 3 Practice: Measuring Stopwords & Noise Reduction Ratio
from nltk.corpus import stopwords
import string

nltk_stopwords = set(stopwords.words('english'))
print("Total NLTK English Stopwords available:", len(nltk_stopwords))

# Use our corpus from Cell 2
raw_tokens = word_tokenize(corpus)
total_raw = len(raw_tokens)

clean_tokens = []
dropped_stopwords = []
dropped_punct = []

for tok in raw_tokens:
    tok_low = tok.lower()
    if tok in string.punctuation or tok in ['``', "''", '—']:
        dropped_punct.append(tok)
    elif tok_low in nltk_stopwords:
        dropped_stopwords.append(tok)
    else:
        clean_tokens.append(tok)

noise_reduction_pct = ((len(dropped_stopwords) + len(dropped_punct)) / total_raw) * 100

print("=" * 70)
print("📊 NOISE REDUCTION & DATASET COMPRESSION REPORT")
print("=" * 70)
print(f"Total Raw Tokens Count     : {total_raw}")
print(f"Stopwords Dropped          : {len(dropped_stopwords)} (e.g. {dropped_stopwords[:6]})")
print(f"Punctuation Dropped        : {len(dropped_punct)}")
print(f"Core Tokens Retained       : {len(clean_tokens)}")
print(f"Noise Reduction Percentage : {noise_reduction_pct:.2f}% 🔥")
print("=" * 70)
print("Core Meaningful Words:")
print(clean_tokens)
````

## 📚 Unit 4: Lemmatization with POS Tagging (Dictionary Roots vs Stemming)
Unlike Stemming, **WordNetLemmatizer** uses the lexical database WordNet to return a valid dictionary word (Lemma).
- Default POS is **Noun** (`pos='n'`).
- Supplying the correct Part-of-Speech (`pos='v'` for verb, `pos='a'` for adjective) allows exact grammatical reduction!

````python
# Unit 4 Practice: Lemmatizer with POS vs Stemmer
from nltk.stem import WordNetLemmatizer, PorterStemmer

lemmatizer = WordNetLemmatizer()
porter = PorterStemmer()

test_cases = [
    ('running', 'v', 'Verb'),
    ('better', 'a', 'Adjective'),
    ('worst', 'a', 'Adjective'),
    ('studies', 'v', 'Verb'),
    ('studies', 'n', 'Noun'),
    ('corpora', 'n', 'Noun (Plural)'),
    ('leaves', 'v', 'Verb'),
    ('leaves', 'n', 'Noun (Plural)'),
    ('was', 'v', 'Verb')
]

print("=" * 90)
print(f"{'Target Word':<12} | {'POS Tag':<16} | {'PorterStemmer':<16} | {'Default Lemma':<16} | {'POS-Aware Lemma':<18}")
print("=" * 90)
for word, pos_code, pos_name in test_cases:
    stem_val = porter.stem(word)
    def_lemma = lemmatizer.lemmatize(word)
    pos_lemma = lemmatizer.lemmatize(word, pos=pos_code)
    print(f"{word:<12} | {pos_name:<16} | {stem_val:<16} | {def_lemma:<16} | {pos_lemma:<18} ✅")
print("=" * 90)
````

## 📊 Unit 5: End-to-End Pipeline & Frequency Distribution (Vocabulary)
Let's assemble a complete, reusable NLP pipeline and visualize the frequency distribution with a clean **ASCII horizontal bar chart**!

````python
# Unit 5 Practice: Complete Pipeline & ASCII Frequency Distribution
from nltk.probability import FreqDist
import string

def nlp_pipeline(text):
    """
    1. Sentence Tokenize
    2. Word Tokenize
    3. Lowercase & Noise Filter (Stopwords & Punctuation)
    4. POS Lemmatization
    5. FreqDist & Vocabulary Extraction
    """
    sentences = sent_tokenize(text)
    raw_tokens = []
    for s in sentences:
        raw_tokens.extend(word_tokenize(s))
        
    stop_words = set(stopwords.words('english'))
    clean_words = []
    for w in raw_tokens:
        w_low = w.lower()
        if w_low not in stop_words and w not in string.punctuation and w.isalnum():
            lemma = lemmatizer.lemmatize(w_low, pos='v')
            clean_words.append(lemma)
            
    freq = FreqDist(clean_words)
    vocab = sorted(list(set(clean_words)))
    
    return {
        'sentences': sentences,
        'raw_tokens': raw_tokens,
        'clean_words': clean_words,
        'vocab': vocab,
        'freq': freq
    }

pipeline_output = nlp_pipeline(corpus)

print("=" * 70)
print("📈 PIPELINE VOCABULARY & CORPUS METRICS:")
print("=" * 70)
print(f"Documents (Sentences)   : {len(pipeline_output['sentences'])}")
print(f"Total Raw Tokens        : {len(pipeline_output['raw_tokens'])}")
print(f"Clean Meaningful Tokens : {len(pipeline_output['clean_words'])}")
print(f"Unique Vocabulary Size  : {len(pipeline_output['vocab'])}")
lexical_diversity = (len(pipeline_output['vocab']) / len(pipeline_output['clean_words'])) * 100 if pipeline_output['clean_words'] else 0
print(f"Lexical Diversity Ratio : {lexical_diversity:.2f}% (Vocabulary Richness)")
print("=" * 70)

print("\n📊 ASCII FREQUENCY DISTRIBUTION BARS (Top Words):")
print("-" * 70)
for word, count in pipeline_output['freq'].most_common(8):
    bar = "█" * (count * 8)
    print(f" {word:<14} | {bar:<28} ({count})")
print("-" * 70)
````

## 🏆 Unit 6: Real-World Practical Challenge (Messy Review Cleaner)
Real-world text data from user reviews contains emojis, typos, contractions, and excitement punctuation.
Let's clean an unstructured review and extract key signals for sentiment analysis!

````python
# Unit 6 Practice: Messy Unstructured Text Cleaner
messy_review = """
WOWWW!! This NLP software version 3.2 is AMAZINGGG & super-fast! 🚀
However, it crashed 2 times today on my MacBook... isn't that annoying? 
Can't wait for bug fixes. Still, I'm giving it 4.5/5 stars because of customer support!
"""

print("=" * 75)
print("🔍 RAW UNSTRUCTURED CUSTOMER REVIEW:")
print(messy_review.strip())
print("=" * 75)

review_res = nlp_pipeline(messy_review)

print("\n✨ EXTRACTED CLEAN KEYWORDS (Ready for Sentiment Model / Classification):")
print(f"Keywords Count : {len(review_res['clean_words'])}")
print(f"Keywords List  : {review_res['clean_words']}")
print(f"Unique Vocab   : {review_res['vocab']}")
print("\nTop 5 Distinct Topic Keywords:")
for kw, count in review_res['freq'].most_common(5):
    print(f" • {kw:<14} (frequency: {count})")
print("=" * 75)
````
