```python
import nltk
nltk.download('punkt_tab')
```

```python
corpus = """Hello! Welcome to Natural Language Processing (NLP) in Python. 
It's fascinating how computers learn to understand human languages, isn't it? 
In 2026, AI models are analyzing millions of texts—from emails to books—every single second. 
Running, ran, and runs all share the same root word, while words like "better" and "good" show how lemmatization works!"""
```

```python
## Tokenization 
## Sentence -->  paragraph
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
## documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization 
## paragraph --> words
## sentence --> words

from nltk.tokenize import word_tokenize
```

```python
## word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize ## it seperate " ' " too
```

```python
for sentence in documents:
    print(wordpunct_tokenize(sentence))
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```
